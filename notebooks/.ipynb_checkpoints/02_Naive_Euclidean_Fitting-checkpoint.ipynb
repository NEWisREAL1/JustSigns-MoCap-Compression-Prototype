# Naive Quaternion Fitting using B-Spline in Euclidean Space

Simple approximation of quaternion sequences by LSPIA using B-Spline on euclidean basis and re-normalization.

```python
%load_ext autoreload
%autoreload 2
```

```python
from copy import deepcopy
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt

from src.utils import prepare_clip, extract_joints
from src.kinematics import forward_kinematics

from src.viz.static import StaticTrajectoriesPLotter
from src.viz.animate import AnimatedSkeletonsPlotter

from src.cagd.fitting import BSplineLSPIAFitter
from src.quaternion import geodesic_distances_rad

from src.compress import compress_clip, decompress_clip
from src.compress import check_antipodal, antipodal_alignment
from src.eval import eval_compression

from src.model import RIGHT_HAND_JOINTS, LEFT_HAND_JOINTS, RIGHT_ARM_JOINTS, LEFT_ARM_JOINTS, CORE_BODY, LOWER_BODY
```

---
## Load Test Clip

```python
CLIP_IDX = 173
```

```python
data = np.load(f"data/npz/clip_{CLIP_IDX}.npz", allow_pickle=True)
clip = data["clip"].item()
clip, clip_name, clip_quats = prepare_clip(clip)

JOINT_NAMES = clip_quats.keys()
NUM_JOINTS  = len(JOINT_NAMES)
NUM_FRAMES  = clip_quats["rootx"].shape[0]
ANIM_STRIDE = np.clip(int(NUM_FRAMES / 100), 1, 5)
```

```python
clip_pos = forward_kinematics(clip_quats)

pl = AnimatedSkeletonsPlotter(title=clip_name, stride=ANIM_STRIDE).apply_defualt_layout()
pl.add_skeleton_frames(clip_pos)
pl.show()
```

---
## Antipodal Alignment (Test)

Make sure no sudden sign flip occured in the data

```python
aligned_quats = deepcopy(clip_quats)

for joint_name, quats in aligned_quats.items():
    if check_antipodal(quats):
        print(f"Sudden sign flip detected: {joint_name}")

        # !!! will fix during compression, leave original data as is
        aligned_quats[joint_name] = antipodal_alignment(quats)
        
        # double check
        if check_antipodal(aligned_quats[joint_name]):
            print(f"-> Could not fix: {joint_name}")
```

```python
aligned_pos = forward_kinematics(aligned_quats)

pl = AnimatedSkeletonsPlotter(title="Antipodal Alignment Consistency Test", stride=ANIM_STRIDE).apply_defualt_layout()
pl.add_skeleton_frames(clip_pos   , name="ground truth"      , joints_color="blue" , bones_color="skyblue", offset=[0.6, 0, 0])
pl.add_skeleton_frames(aligned_pos, name="antipodal aligned" , joints_color="green", bones_color="lightgreen")
pl.show()
```

---
## Compression

```python
fitter = BSplineLSPIAFitter(
    degree              = 2,
    initial_num_cps     = 20,
    time_init_method    = "centripetal",
    knot_init_method    = "time-average",
    cps_init_method     = "sampling-basismax",
    relaxation_schedule = "dynamic",
    centripetal_power   = 0.5,
    # relaxation_factor   = 1.2
)

fit_params = dict(
    max_iter                 = 1000,
    tolerance                = 1e-3,
    converge_iter            = 50,

    dynamic_knot_insertion   = True,
    insert_knot_every        = 100,
    knot_insertion_tolerance = 1e-3,

    # dynamic_reparams         = False,
    # reparams_every           = 10,

    err_agg_func             = np.mean,
    verbose                  = False,
    # print_every              = 25
)

compressed_quats, compress_info = compress_clip(clip_quats, fitter, station_thres_deg=5.0, **fit_params)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(10, 5))

joint = "arm_stretchl"
hist = compress_info["error_history"][joint]
ax.plot(np.arange(len(hist)), hist)

ax.set_title(f"Error over Iterations ({joint})")
ax.set_xlabel("iteration")
ax.set_ylabel("error")
ax.set_yscale("log")
ax.grid()
fig.tight_layout()
plt.show()
```

---
## Regeneration (Decompression)

```python
decompressed_quats = decompress_clip(compressed_quats)
```

```python
decompressed_pos = forward_kinematics(decompressed_quats)

pl = AnimatedSkeletonsPlotter(title=f"Decompressed Clip ({clip_name})", stride=ANIM_STRIDE).apply_defualt_layout()
pl.add_skeleton_frames(clip_pos        , name="ground truth", joints_color="blue", bones_color="skyblue", offset=[0, 0, 0])
pl.add_skeleton_frames(decompressed_pos, name="approximate" , joints_color="red" , bones_color="coral")
pl.show()
```

---
## Evaluation

```python
summary_info, per_joint_info = eval_compression(clip_quats, compressed_quats, compress_info)
df = pd.DataFrame(per_joint_info).T

GRID_JOINTS = [
    [RIGHT_HAND_JOINTS + RIGHT_ARM_JOINTS, CORE_BODY],
    [LEFT_HAND_JOINTS + LEFT_ARM_JOINTS, LOWER_BODY],
]

GRID_JOINTS_LABEL = [
    ["Right Hand & Arm", "Core Body"],
    ["Left Hand & Arm", "Lower Body"],
]
```

### Compression Performance

```python
compression_ratio = summary_info["total_raw_floats"] / summary_info["total_compressed_floats"]
saving_percent = (1 - (1 / compression_ratio)) * 100

print("compression summary")
print(f"-> Total floats used in original data   = {summary_info["total_raw_floats"]:,}")
print(f"-> Total floats used in compressed data = {summary_info["total_compressed_floats"]:,}")
print(f"-> Compression Ratio = {compression_ratio:.6f} ({saving_percent:.2f}% saved)\n")

print("joint types")
print(f"-> {compress_info["num_moving_joints"]} moving joints")
print(f"-> {compress_info["num_nonmoving_joints"]} non-moving joints")
```

```python
fig, axs = plt.subplots(2, 2, figsize=(16, 12), sharey=True, width_ratios=(2, 1))

for i, row in enumerate(axs):
    for j, ax in enumerate(row):
        existing_joints = [ joint for joint in GRID_JOINTS[i][j] if joint in df.index ]
        part_df = df.loc[existing_joints]
        
        bars = ax.bar(np.arange(part_df.shape[0]), part_df["total_compressed_floats"], width=0.8)
        ax.bar_label(bars, padding=3)
        ax.axhline(
            y=df["total_raw_floats"]["rootx"], 
            c="red", linestyle="--", 
            label=f"total floats per joint in original data ({int(df["total_raw_floats"]["rootx"])})"
            )
        
        ax.set_title(f"{GRID_JOINTS_LABEL[i][j]}")
        ax.set_xticks(ticks=np.arange(part_df.shape[0]), labels=part_df.index, rotation=-90)
        ax.set_ylim(0, np.max(df["total_compressed_floats"]) * 1.1)

for ax in axs.flatten():
    ax.legend(loc="upper right")
    ax.grid(axis="y")

fig.suptitle("Floats used by Joints")
fig.tight_layout()
plt.show()
```

### Precision Performance

```python
joint_dists_deg = {}

for joint, de_quats in decompressed_quats.items():
    dists_rad = np.diag(geodesic_distances_rad(de_quats, aligned_quats[joint]))
    dists_deg = np.rad2deg(dists_rad)

    joint_dists_deg[joint] = {}
    joint_dists_deg[joint]["mean_deg"] = np.mean(dists_deg)
    joint_dists_deg[joint]["min_deg"] = np.min(dists_deg)
    joint_dists_deg[joint]["max_deg"] = np.max(dists_deg)
    joint_dists_deg[joint]["std_deg"] = np.std(dists_deg)

dist_df = pd.DataFrame(joint_dists_deg).T
```

```python
fig, axs = plt.subplots(2, 2, figsize=(16, 12), sharey=True, width_ratios=(2, 1))

bar_width = 0.4
offsets = {
    "mean_deg": -bar_width / 2,
    "max_deg": bar_width / 2,
}

for i, row in enumerate(axs):
    for j, ax in enumerate(row):
        existing_joints = [ joint for joint in GRID_JOINTS[i][j] if joint in dist_df.index ]
        part_df = dist_df.loc[existing_joints]
        
        x = np.arange(part_df.shape[0])
        bars_mean = ax.bar(x + offsets["mean_deg"], part_df["mean_deg"], width=bar_width, label="mean")
        bars_max = ax.bar(x + offsets["max_deg"], part_df["max_deg"], width=bar_width, label="max")
        # ax.bar_label(bars_mean, padding=3, fmt="%.2f", fontsize=6)
        # ax.bar_label(bars_max , padding=3, fmt="%.2f", fontsize=6)
        
        ax.set_title(f"{GRID_JOINTS_LABEL[i][j]}")
        ax.set_xticks(ticks=np.arange(part_df.shape[0]), labels=part_df.index, rotation=-90)
        # ax.set_ylim(0, np.max(dist_df[["mean_deg", "min_deg", "max_deg"]].to_numpy()) * 1.1)

for ax in axs.flatten():
    ax.legend()
    ax.grid(axis="y")

fig.suptitle("Mean, Min, and Max Deviation (in Degrees) of Each Joint")
fig.tight_layout()
plt.show()
```

```python
showcase_joint = "forearm_stretchr"
showcase_joint = "handr"

pl = StaticTrajectoriesPLotter(title=f"Positional Trajectory of \"{showcase_joint}\"").apply_defualt_layout()
pl.add_trajectory(clip_pos[showcase_joint]        , name="ground truth", pts_color="blue", line_color="steelblue")
pl.add_trajectory(decompressed_pos[showcase_joint], name="approximate" , pts_color="red" , line_color="crimson")
pl.show()
```

### LSPIA Performance 

```python
fig, axs = plt.subplots(2, 2, figsize=(16, 12), sharey=True, width_ratios=(2, 1))

for i, row in enumerate(axs):
    for j, ax in enumerate(row):
        existing_joints = [ joint for joint in GRID_JOINTS[i][j] if joint in df.index ]
        part_df = df.loc[existing_joints]
        
        bars = ax.bar(np.arange(part_df.shape[0]), part_df["num_iterations"], width=0.8)
        ax.bar_label(bars, padding=3)
        ax.set_title(f"{GRID_JOINTS_LABEL[i][j]}")
        ax.set_xticks(ticks=np.arange(part_df.shape[0]), labels=part_df.index, rotation=-90)
        ax.set_ylim(0, np.max(df["num_iterations"]) * 1.1)

for ax in axs.flatten():
    # ax.legend()
    ax.grid(axis="y")

fig.suptitle("Number of Iterations Used to Fit Each Joint")
fig.tight_layout()
plt.show()
```
